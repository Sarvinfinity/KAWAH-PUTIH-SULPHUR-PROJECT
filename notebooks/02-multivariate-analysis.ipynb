# Multivariate Analysis for Kawah Putih Volcanic Sulphur Dataset

This notebook runs a complete multivariate analysis pipeline for volcanic gas sensor data, including correlation, covariance, pairwise relationships, PCA, feature importance, and node comparisons.

```python
import pandas as pd
from src.multivariate_analysis import (
    compute_correlation_matrix,
    compute_covariance_matrix,
    plot_heatmap,
    plot_pairwise_relationships,
    plot_feature_distributions,
    run_pca,
    plot_pca_variance_explained,
    plot_pca_loadings,
    feature_importance_from_pca,
    feature_importance_from_model,
    compare_node_behaviors,
    summarize_relationships,
]
```

```python
df = pd.read_csv('data/24h_multivariate_volcanic_gas_cleaned.csv', parse_dates=['timestamp'])
df = df.sort_values('timestamp').reset_index(drop=True)
columns = ['SO2', 'H2S', 'CO2', 'humidity', 'temperature']
print('Dataset shape:', df.shape)
print('Columns:', df.columns.tolist())
print('Missing values:
', df[columns].isna().sum())
print('Data types:
', df[columns].dtypes)
```

## Correlation and Covariance
The correlation matrix quantifies linear relationships between sulphur gases, CO2, humidity, and temperature. The covariance matrix indicates joint variability and scale-dependent interactions.

```python
corr = compute_correlation_matrix(df, columns)
cov = compute_covariance_matrix(df, columns)
print(corr)
plot_heatmap(corr, 'Correlation Matrix for Volcanic Gas Features')
plot_heatmap(cov, 'Covariance Matrix for Volcanic Gas Features')
```

## Pair Plots and Distributions
Pair plots reveal joint distributions and potential non-linear structure; histograms show individual variable shape and spread.

```python
plot_pairwise_relationships(df, columns, hue='node_id')
plot_feature_distributions(df, columns)
```

## Principal Component Analysis (PCA)
PCA identifies dominant variance directions and provides a condensed view of feature importance across the dataset.

```python
pca, transformed, variance_ratio = run_pca(df, columns, n_components=3)
print('Explained variance ratio:', variance_ratio)
plot_pca_variance_explained(pca)
plot_pca_loadings(pca, columns, n_components=2)
print(feature_importance_from_pca(pca, columns))
```

## Feature Importance Exploration
We use PCA loadings for unsupervised importance and a Random Forest regressor for supervised importance when a target is specified.

```python
importance = feature_importance_from_model(df, columns, target_column='SO2')
print(importance)
```

## Node 1 vs Node 2 Comparison
Comparing node behavior exposes spatial differences in gas concentration, humidity, and temperature patterns.

```python
compare_node_behaviors(df, columns, node_column='node_id')
```

## Interpretation Notes
- Strong positive SO2-H2S correlation suggests shared volcanic emission dynamics.
- CO2 may show weaker direct correlation with humidity and temperature, indicating partially independent gas transport mechanisms.
- PCA loadings highlight the dominant contribution of SO2 and H2S to the first principal component, while humidity and temperature may shape secondary variance.
- Node comparisons reveal whether one sensor node consistently records higher sulphur levels or distinct meteorological conditions.
